**Exercício 3:**
O objetivo é somar dois números. Por que o Python não consegue fazer isso?

*Dica: O erro é `TypeError: unsupported operand type(s)`. O Python não sabe somar texto com número. Converta a string para número inteiro usando `int()`.*

```python
# CÓDIGO COM ERRO - CONSERTE-ME!
numero1 = int("10")
numero2 = int("5")
soma = numero1 + numero2
print(soma)
```
